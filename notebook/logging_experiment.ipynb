```python
import logging
import os
from datetime import datetime
```

```python
logs_dir = os.path.join(os.getcwd(), "logs")
os.makedirs(logs_dir, exist_ok=True)
```

```python
LOG_FILE = f"{datetime.now().strftime('%m-%d-%Y_%H-%M-%S')}.log"
LOG_FILE
```

```python
LOG_FILE_PATH = os.path.join(logs_dir, LOG_FILE)
```

```python
logging.basicConfig(
    filename=LOG_FILE_PATH,
    format ='[%(asctime)s] %(levelname)s %(name)s (line:%(lineno)d) - %(message)s',
    level=logging.INFO,
)
```

```python
logger = logging.getLogger("DocumentPortal")
```

```python
logger.info("This is an info message.")
```

```python
import structlog
```

```python
# Configure structlog to use the standard library logger
structlog.configure(
    processors=[
        structlog.processors.add_log_level,
        structlog.processors.TimeStamper(fmt="iso"),
        structlog.processors.JSONRenderer() # structlog.dev.ConsoleRenderer()
    ],
    logger_factory=structlog.stdlib.LoggerFactory() 
)
```

```python
logger = structlog.get_logger("DocumentPortal")
```

```python
logger.info("User uploaded a file", user_id=123, filename="report.pdf")
```

```python
logger.error("Failed to process PDF", error="File not found", user_id=123)
```
